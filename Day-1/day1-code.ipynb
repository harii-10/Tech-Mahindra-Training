```python
import math

words=["learn", "build", "sleep", "banana"]
scores=[3.0, 2.0, 0.5, -1.0]

def get_prob(scores, temp):
  scaled=[]
  for score in scores:
    scaled.append(math.exp(score/temp))
  total=sum(scaled)
  prob=[]
  for value in scaled:
    prob.append(value/total)
  return prob

def apply_top_p(words, probabilities, top_p):
    ranked = sorted(zip(probabilities, words), reverse=True)
    kept = []
    running_total = 0
    for probability, word in ranked:
        kept.append(word)
        running_total += probability
        if running_total >= top_p:
            break
    return kept

for temperature in [0.5, 1.5]:
    probs = get_prob(scores, temperature)
    print(f"Temperature {temperature}:")
    for word, probability in zip(words, probs):
        print(f"  {word:<7} {probability:.2%}")

probs = get_prob(scores, 1.0)
print("Top-P 0.9 keeps:", apply_top_p(words, probs, 0.9))


```
